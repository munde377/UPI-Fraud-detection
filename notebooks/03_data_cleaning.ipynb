```python
from pathlib import Path
import pandas as pd
```

```python
# Define projects paths
PROJECT_DIR = Path(r"D:\UPI_Fraud_Detection_Project_ML")
DATA_PATH = PROJECT_DIR / "data" / "upi_fraud_detection_dataset_100k.csv"
CLEANED_PATH = PROJECT_DIR / "data" / "upi_fraud_detection_cleaned.csv"
```

```python
df = pd.read_csv(DATA_PATH)
df["timestamp"] = pd.to_datetime(df["timestamp"], errors="coerce")
```

```python
# keep a copy before cleaning
df_clean = df.copy()
```

```python
# remove exact duplicate rows, if any

df_clean = df_clean.drop_duplicates().copy()
```

```python
# Validate target labels
assert df_clean["is_fraud"].isin([0, 1]).all()
```

```python
# Validate binary Features
binary_cols = ["is_new_device", "is_new_location", "is_weekend"]

for col in binary_cols:
    assert df_clean[col].isin([0, 1]).all(), f"Invalid values  in {col}"
```

```python
# Validate numeric ranges

assert df_clean["amount"].ge(0).all()
assert df_clean["ip_risk_score"].between(0, 100).all()
assert df_clean["device_trust_score"].between(0, 100).all()
assert df_clean["hour"].between(0, 23).all()
assert df_clean["timestamp"].notna().all()
```

```python
# Save cleaned dataset
df_clean.to_csv(CLEANED_PATH, index=False)
```

```python
print("Cleaning completed successfully!")
print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
print("Missing values:", df_clean.isna().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
print("Fraud rate:", round(df_clean["is_fraud"].mean() * 100, 3), "%")
print("Saved to:", CLEANED_PATH)
```
